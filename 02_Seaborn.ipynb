## Basic plots using Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/diamonds.csv')
df.head()
```

```python
sns.scatterplot(x='carat', y='price', data=df, hue='cut')
plt.title('Seaborn Scatter Plot of Carat vs Price')
plt.show()
```

```python
# Line Plot
sns.lineplot(x='carat', y='price', data=df)
plt.title('Seaborn Line Plot of Carat vs Price')
plt.show()
```

## Intermediate Seaborn Plots

```python
# Box Plot
sns.boxplot(x='cut', y='price', data=df)
plt.title('Seaborn Box Plot of Cut vs Price')
plt.show()
```

```python
# Violin Plot
sns.violinplot(x='color', y='price', data=df)
plt.title('Seaborn Violin Plot of Color vs Price')
plt.show()
```

```python
# Count Plot
sns.countplot(x='cut', data=df, hue='color')
plt.title('Seaborn Count Plot of Cut')
plt.show()
```

## Advance Plots

```python
# Heatmap
sns.heatmap(df[['carat','depth','table','price','x','y','z']].corr(),annot=True)
plt.title('Seaborn Heatmap of Correlation')
plt.show()
```

```python
# Pair Plot
sns.pairplot(df, hue='cut')
plt.title('Pair Plot of Diamond Attributes')
plt.show()
```

```python
# Facet Grid
g = sns.FacetGrid(df, col='cut', row='color')
g.map(plt.scatter, 'carat', 'price')
plt.title('Facet Grid of Carat vs Price')
plt.show()
#
```
